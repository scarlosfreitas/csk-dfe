```python
import base64

numero = chave_decimal

# 1. Converte o número inteiro para 8 bytes (64 bits)
bytes_num = numero.to_bytes(8, byteorder='big')

# 2. Codifica para Base64 e remove os caracteres de preenchimento "=" do final
compacto_b64 = base64.urlsafe_b64encode(bytes_num).decode('utf-8').rstrip('=')

print(compacto_b64) 
# Saída: _________w8 (extremamente compacto)
```

```python
ALFABETO = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz"

def int_para_base62(n):
    if n == 0:
        return "0"
        
    resultado = []
    while n > 0:
        n, resto = divmod(n, 62)
        resultado.append(ALFABETO[resto])
        
    # Inverte a lista e junta tudo em uma string
    return ''.join(reversed(resultado))

numero = 18446744073709551615
print(int_para_base62(numero)) 
# Saída: lYGhA16ahyf
```

```python
def reverter_6_bits(numero):
    resultado = 0
    # Como queremos espelhar 6 bits, fazemos um loop de 6 passos
    for _ in range(6):
        # Desloca o resultado 1 posição para a esquerda para abrir espaço
        resultado = resultado << 1
        
        # Pega o bit mais à direita do número original (usando AND 1)
        # e adiciona no resultado (usando OR)
        resultado = resultado | (numero & 1)
        
        # Desloca o número original 1 posição para a direita para ler o próximo bit
        numero = numero >> 1
        
    return resultado
```

```python
print(bin(reverter_6_bits(1)))
print(bin(reverter_6_bits(2)))
print(bin(reverter_6_bits(3)))
```
